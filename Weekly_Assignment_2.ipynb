Question 1: What is a list? Mention three important characteristics of lists.

A list is a collection of elements stored in a single variable in Python


Three characteristics of lists:
1. Ordered: Elements maintain their insertion order.
2. Mutable: We can change, add, or remove elements.
3. Allows duplicates: A list can contain repeated values.

```python
fruits = ["Apple", "Banana", "Mango"]

print(fruits)
```

Question 2: What is the difference between append and extend?

Answer:
Both append() and extend() are used to add elements to a list.

Difference:
- append() adds one element to the list.
- extend() adds multiple elements from another iterable.

```python
# Using append()
list1 = [10, 20, 30]
list1.append(40)

print("Append:", list1)

# Using extend()
list2 = [10, 20, 30]
list2.extend([40, 50])

print("Extend:", list2)
```

Question 3: Write a program to find the largest and smallest values in a list.

Step 1: Create a list of numbers.

Step 2: Use max() to find the largest number.

Step 3: Use min() to find the smallest number.

```python
numbers = [10, 25, 5, 40, 15, 2]

largest = max(numbers)
smallest = min(numbers)

print("Largest number:", largest)
print("Smallest number:", smallest)
```

Question 4: Write a program to calculate the sum and average of numbers stored in a list.
Step 1: Create a list.

Step 2: Use sum() to calculate the total.

Step 3: Divide the sum by the number of elements using len().

```python
numbers = [10, 20, 30, 40, 50]

total = sum(numbers)

average = total / len(numbers)

print("Sum:", total)
print("Average:", average)
```

Question 5: Write a program to remove duplicate values from a list.
Meaning: Duplicate means repeated values.

Example:

[10, 20, 10, 30, 20]

After removing duplicates:

[10, 20, 30]

```python
numbers = [10, 20, 10, 30, 20, 40]

unique_numbers = list(set(numbers))

print("Original List:", numbers)
print("After Removing Duplicates:", unique_numbers)
```

Question 6: What is a tuple? How is it different from a list?

A tuple is a collection of elements stored in a single variable. A tuple is immutable, which means we cannot change its elements after creation.

| List | Tuple |
|---|---|
| Uses [] brackets | Uses () brackets |
| Mutable | Immutable |
| Elements can be changed | Elements cannot be changed |
| Example: [10, 20, 30] | Example: (10, 20, 30) |

```python
fruits = ("Apple", "Banana", "Mango")

print(fruits)
```

Question 7: What is tuple unpacking? Explain with a real-life

Tuple unpacking means assigning tuple elements to different variables.

Real-life example:

Suppose a student has a name, age and course.

Here:
- name stores Kavana.
- age stores 21.
- course stores Python.
This process is called tuple unpacking.

```python
student = ("Kavana", 21, "Python")

name, age, course = student

print("Name:", name)
print("Age:", age)
print("Course:", course)
```

Question 9: Write a program to find common elements between two collections.

Meaning: Find the elements that are present in both collections.

Example:

Collection 1 = [10, 20, 30, 40]

Collection 2 = [30, 40, 50, 60]

Common elements = [30, 40]

```python
list1 = [10, 20, 30, 40]
list2 = [30, 40, 50, 60]

common = set(list1).intersection(set(list2))

print("Common Elements:", common)
```

Question 10: What is a dictionary? Explain keys and values.

A dictionary is a collection of data stored in key-value pairs.
- Key: Used to identify the data.
- Value: The actual data associated with the key.

```python
student = {
    "name": "Kavana",
    "age": 21,
    "course": "Python"
}

print("Student Name:", student["name"])
print("Student Age:", student["age"])
print("Student Course:", student["course"])
```
